# EDA

```python
# IMPORTS NECESARIOS
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# CARGA DEL ARCHIVO

ruta = r'C:\RUTA\AL\ARCHIVO.csv'

df = pd.read_csv(
    ruta,
    sep=';'
)

print("Archivo cargado correctamente.")
```

```python
# VISTA GENERAL

print(f"Dimensiones: {df.shape}")

df.head()
```

```python
# ESTRUCTURA Y TIPOS DE DATOS

df.info()

print("\nCantidad de columnas por tipo de dato:")
print(df.dtypes.value_counts())
```

```python
# VALORES NULOS

nulos = df.isnull().sum()

pct_nulos = (nulos / len(df)) * 100

resumen_nulos = pd.DataFrame({
    'nulos': nulos,
    'pct': pct_nulos.round(2)
})

resumen_nulos = (
    resumen_nulos[
        resumen_nulos['nulos'] > 0
    ]
    .sort_values('pct', ascending=False)
)

resumen_nulos
```

```python
# DUPLICADOS

print(f"Filas duplicadas: {df.duplicated().sum()}")
```

```python
# DUPLICADOS POR CLAVE / ID
# Reemplazar 'COLUMNA_ID' por la clave correspondiente

print(
    f"Duplicados por ID: "
    f"{df.duplicated(subset=['COLUMNA_ID']).sum()}"
)
```

```python
# ESTADÍSTICAS DESCRIPTIVAS
# VARIABLES NUMÉRICAS

num_cols = df.select_dtypes(include=np.number).columns

df[num_cols].describe().T
```

```python
# ESTADÍSTICAS DESCRIPTIVAS
# VARIABLES CATEGÓRICAS

df.describe(include=['object']).T
```

```python
# VARIABLES CATEGÓRICAS - CARDINALIDAD

cat_cols = df.select_dtypes(include='object').columns

for col in cat_cols:

    n_unique = df[col].nunique()

    print(f"{col}: {n_unique} valores únicos")

    if n_unique <= 15:
        print(
            df[col]
            .value_counts(normalize=True, dropna=False)
            .round(3) * 100
        )

    print("-" * 40)
```

```python
# DISTRIBUCIÓN DE VARIABLES NUMÉRICAS

df[num_cols].hist(
    figsize=(14, 10),
    bins=30
)

plt.tight_layout()
plt.show()
```

```python
# OUTLIERS

for col in num_cols:

    plt.figure(figsize=(8, 3))

    sns.boxplot(
        data=df,
        x=col
    )

    plt.title(f'Boxplot de {col}')
    plt.show()
```

## RESUMEN GENERAL 

```python
# RESUMEN GENERAL DE DATASETS

datasets = {}

resumen = []

for nombre, df in datasets.items():

    resumen.append({
        'Dataset': nombre,
        'Filas': df.shape[0],
        'Columnas': df.shape[1],
        'Nulos': df.isnull().sum().sum(),
        'Duplicados': df.duplicated().sum()
    })

resumen_general = pd.DataFrame(resumen)

resumen_general
```

```python
# INVENTARIO GENERAL DE COLUMNAS

inventario = []

for nombre, df in datasets.items():

    for col in df.columns:

        inventario.append({
            'Dataset': nombre,
            'Columna': col,
            'Tipo': str(df[col].dtype),
            'Valores únicos': df[col].nunique(),
            'Nulos': df[col].isnull().sum(),
            '% Nulos': round(df[col].isnull().mean() * 100, 2)
        })

inventario_columnas = pd.DataFrame(inventario)

inventario_columnas
```
